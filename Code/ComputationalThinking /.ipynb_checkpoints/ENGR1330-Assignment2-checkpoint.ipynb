```python
# B)
Sally = {
    "January":670,
    "February":689,
    "March":685,
}
Dominic = {
    "January":720,
    "February":714,
    "March":725,
}
Herbert = {
    "January":801,
    "February":799,
    "March":804,
}
Wendy = {
    "January":745,
    "February":741,
    "March":751,
}

def avg(x,y,z):
    average = (x+y+z)/3
    return average
    
    
```

```python
SallyAvg = avg(Sally["January"],Sally["February"],Sally["March"])
DominicAvg = avg(Dominic["January"],Dominic["February"],Dominic["March"])
HerbertAvg = avg(Herbert["January"],Herbert["February"],Herbert["March"])
WendyAvg = avg(Wendy["January"],Wendy["February"],Wendy["March"])

print(f"Sally: {SallyAvg:.2f}, Dominic: {DominicAvg:.2f}, Herbert: {HerbertAvg:.2f}, Wendy: {WendyAvg:.2f}")
```
