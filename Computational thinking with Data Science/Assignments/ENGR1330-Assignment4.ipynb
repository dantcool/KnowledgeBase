## Question 2:

Create a function, named `mypolynomial` that takes the single positional argument `x` as input and produces an output value `y` for the equation

$$ y = -5x^5 +4sin(x)^4 -3x^3 +2cos(x)^2 -x $$
    
Use the function to create a pair of lists of x raging from -10 to 10 (inclusive) in steps of 0.5, and the corresponding `y` values

```python
# build your script here
import math 
from prettytable import PrettyTable
import numpy as np

def mypolynomial(a):
    y = -5*a**5 + 4*math.sin(a)**4 - 3*a**3+2*math.cos(a)**2-a
    return y 

xlist=[]
ylist=[]
table=PrettyTable(['x','y'])

x=-10
while x <=10:
    xlist.append(x)
    ylist.append(mypolynomial(x))
    table.add_row([x,mypolynomial(x)])
    x += .5

print(table)





```

```python
# script goes here
import math

def vofCylinder(r,h):
    volume = math.pi*(r**2)*h
    print (volume)
    return volume

def vofSphere(r):
    volume = 4/3*(math.pi*r**3)
    print (volume)
    return volume

def vofCone(r,h):
    volume = (math.pi*(r**2)*h)/3
    print (volume)
    return volume 

def vofRectangle(w,h,l):
    volume = w*h*l
    print (volume)
    return volume
    
print("Please choose the shape of the object")
ans = input("Enter 1 for Cylinder, 2 for Sphere, 3 for Cone, or 4 for Rectangular Box")
try:
    ans = int(ans)
    
except:
    print("make sure answer is a number")

if ans < 1 or ans > 4:
    print('make sure choice is between 1 and 4')
else:
    match ans:
    
        case 1:
            r = float(input("Enter Radius"))  
            h = float(input("Enter height"))
            vofCylinder(r,h)
        case 2:
            r = float(input("Enter radius"))
            vofSphere(r)
        case 3:
            r = float(input("Enter Radius"))  
            h = float(input("Enter height"))
            vofCone(r,h)
        case 4:
            w = float(input("Enter width"))
            h = float(input("Enter height"))
            l = float(input("Enter length"))
            vofRectangle(w,h,l)
```
