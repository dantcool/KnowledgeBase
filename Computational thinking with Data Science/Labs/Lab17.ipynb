## EXAMPLE 1. Code Generation
Generative AI can create Python functions from simple natural language prompts. This is useful when prototyping or learning new concepts.

```python
# Example 1: Generate a function to compute factorial
# Prompt: "Write a Python function to compute the factorial of a number"
def factorial(n):
    if n == 0 or n == 1:
        return 1
    return n * factorial(n-1)

# Test
print(factorial(5))  # Output should be 120
```

## EXAMPLE 2. Code Refactoring
Generative AI can refactor messy or unoptimized code into cleaner, more efficient versions.

```python
# Example 2: Refactor a loop into a list comprehension
# Original code:
squares = []
for i in range(10):
    squares.append(i * i)

# Refactored:
squares = [i * i for i in range(10)]
print(squares)
```

## EXAMPLE 3. Debugging
Generative AI can help find bugs and suggest fixes in your code.

```python
# Example 3: Debug a function that adds two lists element-wise
# Buggy code:
def add_lists(a, b):
    return [a + b for a, b in zip(a, b)]  # Variable name conflict

# Corrected code:
def add_lists(a_list, b_list):
    return [x + y for x, y in zip(a_list, b_list)]

# Test
print(add_lists([1, 2], [3, 4]))  # Output: [4, 6]
```

## Example 4: Generate Data Visualization
Generative AI can quickly generate plots using libraries like matplotlib or seaborn based on your data description.

```python
# Prompt: "Plot a bar chart showing sales over 5 months"
import matplotlib.pyplot as plt

months = ['Jan', 'Feb', 'Mar', 'Apr', 'May']
sales = [200, 250, 300, 280, 320]

plt.bar(months, sales)
plt.title('Monthly Sales')
plt.xlabel('Month')
plt.ylabel('Sales')
plt.show()
```

## Example 5: Interact with APIs
Generative AI can help you create code to consume public APIs, saving time on documentation lookup.

```python
# Prompt: "Get current weather from OpenWeatherMap API"
import requests

api_key = "your_api_key_here"
city = "London"
url = f"https://api.openweathermap.org/data/2.5/weather?q={city}&appid={api_key}&units=metric"

response = requests.get(url)
data = response.json()

if response.status_code == 200:
    print(f"Current temperature in {city}: {data['main']['temp']}°C")
else:
    print("Error fetching weather data")
```
