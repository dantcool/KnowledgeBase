## <font color=purple>How do we approach a problem?</font>

### The Scientific Method:

__1. Observation__

   Formulation of a question
   
__2. Hypothesis__

   A conjecture that may explain observed behavior. Falsifiable by an experiment whose outcome conflicts with predictions deduced from the hypothesis
   
__3. Prediction__

   How the experiment should conclude if hypothesis is correct 
   
__4. Testing__

   Experimental design, and conduct of the experiment.
   
__5. Analysis__

   Interpretation of experimental results

### The Computational (Programming) Method:

__1. Define the problem__

   Explicitly state the problem
   
__2. Gather information__

    1.	Input information
    2.	Governing equations or principles, and 
    3.	The required output information.

   
__3. Generate and evaluate potential solutions__

   Work a sample problem by-hand for developing and testing a generic solution 
   
__4. Testing__

   Refine and implement a solution. Develop a generic solution method (coding).
   
__5. Verification__

   Verify and test the solution. Test the generic solution against the by-hand example, then apply to the real problem

## CCMR (Copy, Cite, Modify, Run) or How an Engineering Data Scientist Makes a Living?

### A lot of the problems we will encounter have already been solved, or at least their close analogs. Remember "pattern recognition"!

 
 
 
__Copy__

Find a solution to your problem from some online example: SourceForge, StackOverflow, GeeksFor-Geeks, DigitalOcean, etc., or god forbid get it from a cellulose based device!
   
__Cite__

Cite the original source. In general a citation will look like one in a book, but a URL to the source is a sufficient start.

   
__Modify__

Modify the original cited work for your specific needs. Note the changes in the code using comment statements. 
   
__Run__

Apply the modified code to the problem of interest.
